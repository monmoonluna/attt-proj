```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

X_train = pd.read_csv("output/CICDDoS2019/X_train.csv")
X_test = pd.read_csv("output/CICDDoS2019/X_test.csv")
y_train = pd.read_csv("output/CICDDoS2019/y_train.csv")["label"]
y_test = pd.read_csv("output/CICDDoS2019/y_test.csv")["label"]

selected = [
    "Destination Port", "Total Backward Packets", "Total Length of Fwd Packets",
    "Total Length of Bwd Packets", "Fwd Packet Length Max", "Fwd Packet Length Mean",
    "Bwd Packet Length Min", "Bwd Packet Length Mean", "Bwd Packet Length Std",
    "Flow IAT Max", "Fwd IAT Std", "Fwd IAT Min", "Fwd Header Length",
    "Fwd Packets/s", "Max Packet Length", "Packet Length Variance",
    "URG Flag Count", "CWE Flag Count", "Avg Fwd Segment Size",
    "Subflow Bwd Packets", "act_data_pkt_fwd", "Active Mean", "Active Min",
    "Idle Mean", "Idle Max",
]

clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train[selected], y_train)

importances = pd.Series(clf.feature_importances_, index=selected).sort_values(ascending=False)
print(importances)

# Thử chỉ dùng top-3 đặc trưng quan trọng nhất - nếu vẫn ~0.99 thì rất đáng ngờ
top3 = importances.head(3).index.tolist()
clf3 = DecisionTreeClassifier(random_state=42).fit(X_train[top3], y_train)
from sklearn.metrics import f1_score
print("Top-3 only:", top3, "F1 =", f1_score(y_test, clf3.predict(X_test[top3]), average="macro"))
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import f1_score

X_train = pd.read_csv("output/CICDDoS2019/X_train.csv")
X_test = pd.read_csv("output/CICDDoS2019/X_test.csv")
y_train = pd.read_csv("output/CICDDoS2019/y_train.csv")["label"]
y_test = pd.read_csv("output/CICDDoS2019/y_test.csv")["label"]

def ablation_drop(name, selected, drop_col):
    without_col = [c for c in selected if c != drop_col]
    for label, cols in [(f"{name} - đủ {len(selected)} (có {drop_col})", selected),
                         (f"{name} - bỏ {drop_col} (còn {len(without_col)})", without_col)]:
        clf = DecisionTreeClassifier(random_state=42).fit(X_train[cols], y_train)
        f1 = f1_score(y_test, clf.predict(X_test[cols]), average="macro")
        print(f"{label}: F1 = {f1:.4f}")

filter_mi_selected = [
    "Destination Port", "Flow Duration", "Total Length of Fwd Packets",
    "Fwd Packet Length Max", "Fwd Packet Length Min", "Fwd Packet Length Mean",
    "Flow Bytes/s", "Flow IAT Mean", "Flow IAT Max", "Fwd Packets/s",
    "Min Packet Length", "Max Packet Length", "Packet Length Mean",
    "Packet Length Variance", "Average Packet Size", "Avg Fwd Segment Size",
    "Subflow Fwd Bytes", "Init_Win_bytes_forward", "Inbound",
]
hybrid2_selected = [
    "Destination Port", "Flow Duration", "Total Length of Fwd Packets",
    "Fwd Packet Length Max", "Fwd Packet Length Min", "Fwd Packet Length Mean",
    "Flow Bytes/s", "Flow IAT Mean", "Flow IAT Std", "Flow IAT Max",
    "Fwd Packets/s", "Min Packet Length", "Max Packet Length",
    "Packet Length Mean", "URG Flag Count", "Average Packet Size",
    "Avg Fwd Segment Size", "Subflow Fwd Bytes", "Init_Win_bytes_forward", "Inbound",
]

ablation_drop("Filter-MI", filter_mi_selected, "Inbound")
ablation_drop("Hybrid2", hybrid2_selected, "Inbound")

# Kiểm tra thêm: phân bố Inbound theo nhãn - xem có phải gần như hằng số theo từng lớp
df_check = X_train[["Inbound"]].copy()
df_check["label"] = y_train.values
print(df_check.groupby("label")["Inbound"].value_counts(normalize=True))
```

```python
import json
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import f1_score

d = "output/CICDDoS2019"
X_train = pd.read_csv(f"{d}/X_train.csv"); X_test = pd.read_csv(f"{d}/X_test.csv")
y_train = pd.read_csv(f"{d}/y_train.csv")["label"]; y_test = pd.read_csv(f"{d}/y_test.csv")["label"]
selected = json.load(open(f"{d}/fs_ga_rf/selected_features.json", encoding="utf-8"))["selected_features"]
print("6 đặc trưng GA chọn:", selected)

def f1_of(cols, clf):
    clf.fit(X_train[cols], y_train)
    return f1_score(y_test, clf.predict(X_test[cols]), average="macro")

for name, make in [("RF", lambda: RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)),
                   ("DT", lambda: DecisionTreeClassifier(random_state=42))]:
    print(f"\n=== {name} ===")
    print(f"Đủ {len(selected)} đặc trưng: F1 = {f1_of(selected, make()):.4f}")
    for feat in selected:
        rest = [c for c in selected if c != feat]
        print(f"  Bỏ '{feat}': F1 = {f1_of(rest, make()):.4f}")
```

```python
import json
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import f1_score

d = "output/CICDDoS2019"
X_train = pd.read_csv(f"{d}/X_train.csv"); X_test = pd.read_csv(f"{d}/X_test.csv")
y_train = pd.read_csv(f"{d}/y_train.csv")["label"]; y_test = pd.read_csv(f"{d}/y_test.csv")["label"]

models = {
    "DT": lambda: DecisionTreeClassifier(random_state=42),
    "RF": lambda: RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    "XGB": lambda: XGBClassifier(n_estimators=200, max_depth=6, random_state=42, n_jobs=-1, verbosity=0),
}
to_add = {"fs_filter_rf": "URG Flag Count", "fs_hybrid1_pso_rf": "Destination Port"}

for name, extra in to_add.items():
    feats = json.load(open(f"{d}/{name}/selected_features.json", encoding="utf-8"))["selected_features"]
    for label, cols in [("gốc", feats), (f"+{extra}", feats + [extra])]:
        res = []
        for m, make in models.items():
            clf = make().fit(X_train[cols], y_train)
            res.append(f"{m}={f1_score(y_test, clf.predict(X_test[cols]), average='macro'):.4f}")
        print(f"{name:<18} {label:<22} ({len(cols):>2} đặc trưng): " + "  ".join(res))
```

```python
import json, pandas as pd
from xgboost import XGBClassifier
from sklearn.metrics import f1_score

d = "output/CICDDoS2019"
X_train = pd.read_csv(f"{d}/X_train.csv"); X_test = pd.read_csv(f"{d}/X_test.csv")
y_train = pd.read_csv(f"{d}/y_train.csv")["label"]; y_test = pd.read_csv(f"{d}/y_test.csv")["label"]
feats = json.load(open(f"{d}/fs_hybrid1_pso_rf/selected_features.json", encoding="utf-8"))["selected_features"]

orders = {"thứ tự trong JSON": feats,
          "thứ tự cột gốc": [c for c in X_train.columns if c in feats],
          "đảo ngược": feats[::-1]}
for name, cols in orders.items():
    for run in (1, 2):
        clf = XGBClassifier(n_estimators=200, max_depth=6, random_state=42, n_jobs=-1, verbosity=0).fit(X_train[cols], y_train)
        f1 = f1_score(y_test, clf.predict(X_test[cols]), average="macro")
        print(f"{name:<20} lần {run}: F1 = {f1:.4f}")
```
